```python
up_vil_unfiltered = up_vil_unfiltered
up_vil = up_vil_unfiltered.replace(r'^\s*$', np.nan, regex=True)
up_vil = up_vil.dropna(subset=['vilnam_soi'])
up_vil = up_vil.dropna(subset=['vilname11'])
vil_name= up_vil['vilname11'].to_list()
vil_name
```

```python
up_vil = up_vil.loc[~up_vil['vilcode11'].isna()]
up_vil
```

```python
up_vil = up_vil.loc[up_vil["stname"] == "UTTAR PRADESH"]
up_vil["vilnam_soi"] = up_vil["vilnam_soi"].str.upper()
up_vil = up_vil[~up_vil["vilnam_soi"].str.contains("FOREST", na=False)]
up_vil = up_vil[~up_vil["vilnam_soi"].str.contains("RF", na=False)]
up_vil = up_vil[~up_vil["vilnam_soi"].str.contains("RESERVED FOREST", na=False)]
#up_vil = up_vil[~up_vil["vilnam_soi"].str.contains("D.P.F", na=False)]
up_vil = up_vil[~up_vil["vilnam_soi"].str.contains(" RF", na=False)]
up_vil = up_vil[~up_vil["vilnam_soi"].str.contains("HILL", na=False)]
up_vil = up_vil[~up_vil["vilnam_soi"].str.contains("R F", na=False)]
up_vil
```

```python
up_vil['dtcode11'] = up_vil['dtcode11'].astype(str).apply(lambda x: x.zfill(2))
up_vil['block_lgd'] = up_vil['block_lgd'].astype(float).astype(int).astype(str).apply(lambda x: x.zfill(5))
up_vil['object_id'] = up_vil.apply(lambda row: f"21-{row['dtcode11']}-{row['block_lgd']}", axis=1)
up_vil
```

```python
up_vil = up_vil.drop(['object_id','remark', 'ac_no'],axis=1)
up_vil.to_file(r'D:\CDL\flood-data-ecosystem-UP\Maps\up_ids-drr_shapefiles\up_villages_final_filtered.geojson')
```

```python
up_urban = gpd.read_file(r'D:\CDL\flood-data-ecosystem-UP\Maps\up_ids-drr_shapefilesup_villages_final_filtered.geojson')
up_urban
```

```python
up_urban = up_urban.dropna(subset=['vilnam_soi'])
up_urban = up_urban.loc[up_urban["stcode11"] == "09"]
up_urban["vilnam_soi"] = up_urban["vilnam_soi"].str.upper()
up_urban = up_urban[~up_urban["vilnam_soi"].str.contains("FOREST", na=False)]
up_urban = up_urban[~up_urban["vilnam_soi"].str.contains("RF", na=False)]
up_urban = up_urban[~up_urban["vilnam_soi"].str.contains("RESERVED FOREST", na=False)]
up_urban = up_urban[~up_urban["vilnam_soi"].str.contains(" RF", na=False)]
up_urban = up_urban[~up_urban["vilnam_soi"].str.contains("HILL", na=False)]
up_urban = up_urban[~up_urban["vilnam_soi"].str.contains("R F", na=False)]

up_urban
```

```python
# https://en.wikipedia.org/wiki/List_of_urban_agglomerations_in_Uttar_Pradesh
urban_filter = ['kanpur','lucknow','ghaziabad','agra','varanasi','meerut','prayagraj','bareilly','aligarh','gorakhpur','ayodhya', 'jhansi', 'muzaffarnagar', 
                'mathura','rampur','shahjahanpur','Farrukhabad','mirzapur','bulandshahr','hardoi','orai','sitapur','modinagar','lakhimpur','hathras','banda','pilibhit','khurja','gonda','mainpuri','etah','ghazipur','sultanpur',
                'azamgarh','ballia' ,'(NPP)','(M Corp.)','(NP)','(OG)','(CB)','(CT)']
filtered_gdf = up_urban[up_urban['vilname11'].str.contains('|'.join(urban_filter), na=False)]
filtered_gdf
```

```python
merged = filtered_gdf.merge(up_vil[['vilcode11']],left_on='vilcode11', right_on='vilcode11',how='left')

merged

```

```python
merged = merged.drop('vilcode11',axis=1)
final_gdf = gpd.GeoDataFrame(merged,geometry='geometry')
final_gdf.to_file(r'D:\CDL\flood-data-ecosystem-UP\Maps\up_ids-drr_shapefiles\up_urban_final.geojson')
```
